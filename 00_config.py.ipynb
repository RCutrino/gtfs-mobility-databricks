```python
import json
import os
from datetime import datetime
```

```python
# ------------------------------------------------------------------------------
# 1. GLOBAL CONFIG
# ------------------------------------------------------------------------------

BASE_PATH = "/Volumes/workspace/default/gtfs-mobility/progetto_databricks"

CITIES = {
    "roma": {
        "folder": "roma",
        "url": "https://dati.comune.roma.it/catalog/dataset/a7dadb4a-66ae-4eff-8ded-a102064702ba/resource/266d82e1-ba53-4510-8a81-370880c4678f/download/rome_static_gtfs.zip"},
    "milano": {
        "folder": "milano",
        "url": "https://dati.comune.milano.it/gtfs.zip"},
    "torino": {
        "folder": "torino",
        "url": "https://www.gtt.to.it/open_data/gtt_gtfs.zip"},
    "napoli": {
        "folder": "napoli",
        "url": "http://www.anm.it/google/google-transit.zip"},
    "bologna": {
        "folder": "bologna",
        "url": "https://solweb.tper.it/web/tools/open-data/open-data-download.aspx?source=solweb.tper.it&filename=gommagtfsbo&version=20261006&format=zip"},
    "firenze": {
        "folder": "firenze",
        "url": "https://files.mobilitydatabase.org/tld-6959/tld-6959-202604010041/tld-6959-202604010041.zip"}
}

GTFS_TABLES = [
    "stops",
    "routes", 
    "trips",
    "stop_times",
    "calendar",
    "calendar_dates"]
```

```python
# ------------------------------------------------------------------------------
# 2. GENERATE runtime_config.json
# cities_without_calendar is empty here — 00_setup.py will populate it.
# ------------------------------------------------------------------------------

CONFIG_FILE = f"{BASE_PATH}/runtime_config.json"
now = datetime.now().isoformat(timespec="seconds")

runtime_config = {
    "metadata": {
        "created_at": now,
        "updated_at": now,
        "updated_by": "00_config.py",
        "version":    "v2"},
    "base_path":                BASE_PATH,
    "cities":                   CITIES,
    "gtfs_tables":              GTFS_TABLES,
    "cities_without_calendar":  []}

# Preserve created_at if config already exists
if os.path.exists(CONFIG_FILE):
    with open(CONFIG_FILE, "r") as f:
        existing = json.load(f)
    runtime_config["metadata"]["created_at"] = existing["metadata"].get("created_at", now)
    print(f"  ℹ️  existing config found — preserving created_at: {runtime_config['metadata']['created_at']}")

with open(CONFIG_FILE, "w") as f:
    json.dump(runtime_config, f, indent=2)

print(f"✓ runtime_config.json generated → {CONFIG_FILE}")
print(f"  created_at:              {runtime_config['metadata']['created_at']}")
print(f"  updated_at:              {runtime_config['metadata']['updated_at']}")
print(f"  updated_by:              {runtime_config['metadata']['updated_by']}")
print(f"  version:                 {runtime_config['metadata']['version']}")
print(f"  cities:                  {list(CITIES.keys())}")
print(f"  gtfs_tables:             {GTFS_TABLES}")
print(f"  cities_without_calendar: [] — will be populated by 00_setup.py")
```
