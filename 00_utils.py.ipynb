```python
# Utility functions for reading and updating runtime_config.json.
# %run this notebook from any other notebook to access load_config()
# and update_config() — no side effects, no file generation.
```

```python
import json
import os
from datetime import datetime
```

```python
CONFIG_FILE = "/Volumes/workspace/default/gtfs-mobility/progetto_databricks/runtime_config.json"
```

```python
def load_config():
    """Reads runtime_config.json and returns the config dictionary."""
    with open(CONFIG_FILE, "r") as f:
        return json.load(f)

def update_config(key, value, updated_by):
    """
    Updates a key in runtime_config.json.
    Automatically updates metadata.updated_at and metadata.updated_by.
    """
    config = load_config()
    config[key] = value
    config["metadata"]["updated_at"] = datetime.now().isoformat(timespec="seconds")
    config["metadata"]["updated_by"] = updated_by
    with open(CONFIG_FILE, "w") as f:
        json.dump(config, f, indent=2)
    print(f"✓ config updated → {key}: {value} (by {updated_by})")

print("✓ utils loaded: load_config(), update_config()")
```
